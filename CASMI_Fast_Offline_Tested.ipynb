# CASMI — Fast Offline Retrieval
Run all cells in order. Internet OFF. Attach official competition files and the existing RDKit wheel dataset. CPU is sufficient. Output: `/kaggle/working/submission.csv`.

## Changes and limits
- Uses IDs from the current test file; no public sample dependency.
- Streams training spectra; retains the mass-matched candidates without an arbitrary top-80 cut.
- Optional Numba compilation accelerates greedy matching; tested Python fallback needs no extra installation.
- Tautomer canonicalization is deferred until ranking, avoiding work for candidates that are never submitted. Scores are aggregated by input SMILES before final identity deduplication; this can differ from aggregating evidence across tautomers first.
- Supports formula-based adducts, ignores malformed spectra, and logs fallback use. When no usable candidate exists, a frequency-based prior from real training structures supplies valid guesses. This is a weak fallback, not an identification.
- No model training, external downloads, global exception suppression, or fake success messages. Missing required input/dependencies still produce actionable errors.

## Reference review
The supplied de novo example uses unshifted training targets and incomplete generation context; it is not included. The two supplied multi-channel rankers require `rank_train.npz`, `fp_bits.npy`, `coco_meta.pkl`, `coco_mass.npy`, `coco_fp.npy`, and FPNet checkpoints for their complete pipeline. Those assets were not supplied, so their ranker and reported scores are not reproduced here. Ideas retained: mass-constrained candidate search, multi-spectrum evidence, offline execution and optional compiled matching.

The reference score claims are unverified. This notebook is a fast retrieval baseline, not a demonstrated best-accuracy solution. Structures absent from the training library cannot be recovered. Kaggle hidden-run acceptance is not guaranteed.

## 3. Import libraries and define adduct masses

```python
import argparse
import json
import time
from pathlib import Path
from functools import lru_cache

import numpy as np
import pandas as pd
import pyarrow.parquet as pq
from rdkit import Chem, rdBase, RDLogger
from rdkit.Chem import Descriptors
from rdkit.Chem.MolStandardize import rdMolStandardize

RDLogger.DisableLog('rdApp.*')
TAUTOMER = rdMolStandardize.TautomerEnumerator()
# Ionic mass added to neutral M, for singly charged test adducts.
ADDUCT_DELTA = {
    '[M+H]+': 1.007276466621, '[M-H]-': -1.007276466621,
    '[M+Na]+': 22.989218, '[M+K]+': 38.963158,
    '[M+NH4]+': 18.033823, '[M+Cl]-': 34.969402,
    '[M+CH2O2-H]-': 44.998201, '[M+HCOO]-': 44.998201,
}

import re
from collections import Counter
DIAGNOSTICS = Counter()
def number(value):
    try: return float(value)
    except (TypeError, ValueError): return float('nan')
def adduct_text(value):
    return str(value).replace(' ', '').replace('−','-')
def neutral_mass(mz, adduct):
    mz = number(mz)
    if not np.isfinite(mz) or mz <= 0: return np.nan
    adduct = adduct_text(adduct)
    if adduct in ADDUCT_DELTA: return mz - ADDUCT_DELTA[adduct]
    # Ionic mass = neutral atom masses - signed charge * electron mass.
    match = re.fullmatch(r'\[(\d*)M((?:[+-][A-Za-z0-9]+)*)\](\d*)([+-])', adduct)
    if not match: return np.nan
    mult, terms, charge, sign = match.groups()
    mult, charge = int(mult or 1), int(charge or 1)
    if mult < 1 or charge < 1: return np.nan
    delta = 0.0
    aliases = {'ACN':'C2H3N','FA':'CH2O2','Hac':'C2H4O2','Ac':'C2H3O2'}
    table = Chem.GetPeriodicTable()
    for op, term in re.findall(r'([+-])([^+-]+)', terms):
        coefficient = re.match(r'^(\d*)(.*)$',term)
        count, formula = coefficient.groups(); formula = aliases.get(formula,formula)
        atoms = re.findall(r'([A-Z][a-z]?)(\d*)',formula)
        if ''.join(a+b for a,b in atoms) != formula or not atoms: return np.nan
        try: mass = sum(table.GetMostCommonIsotopeMass(a)*int(b or 1) for a,b in atoms)
        except Exception: return np.nan
        delta += (1 if op=='+' else -1)*int(count or 1)*mass
    signed_charge = charge*(1 if sign=='+' else -1)
    result = (mz*charge - delta + signed_charge*0.000548579909065)/mult
    return result if result > 0 else np.nan

def mode_text(value, adduct):
    s = str(value).lower().strip()
    if s in {'positive','pos','+','1','1.0'}: return 'positive'
    if s in {'negative','neg','-','-1','-1.0'}: return 'negative'
    a = adduct_text(adduct)
    return 'positive' if a.endswith('+') else ('negative' if a.endswith('-') else 'unknown')
```

## 4. Load and check data

## 5. Define chemistry and spectral matching functions

```python
@lru_cache(maxsize=50000)
def identity(smiles):
    try:
        mol = Chem.MolFromSmiles(smiles)
        if mol is None:
            return None
        mol = TAUTOMER.Canonicalize(mol)
        key = Chem.MolToInchiKey(mol).split('-')[0]
        if len(key) != 14:
            return None
        return key, Chem.MolToSmiles(mol, isomericSmiles=False)
    except Exception:
        return None

def prepare_peaks(mzs, intensities, precursor, losses=False):
    empty = (np.array([]), np.array([]))
    try:
        mz = np.asarray(mzs, dtype=float)
        intensity = np.asarray(intensities, dtype=float)
    except (TypeError, ValueError):
        DIAGNOSTICS['invalid_peak_arrays'] += 1
        return empty
    if mz.ndim != 1 or intensity.ndim != 1 or mz.shape != intensity.shape:
        DIAGNOSTICS['invalid_peak_arrays'] += 1
        return empty
    precursor = number(precursor)
    if losses and not np.isfinite(precursor): return empty
    keep = np.isfinite(mz) & np.isfinite(intensity) & (mz > 0) & (intensity > 0)
    if np.isfinite(precursor): keep &= np.abs(mz - precursor) > 1.5
    mz, intensity = mz[keep], intensity[keep]
    if len(mz) == 0:
        return np.array([]), np.array([])
    keep = np.argsort(intensity)[-128:]
    mz, intensity = mz[keep], np.sqrt(intensity[keep])
    if losses:
        mz = precursor - mz
        keep = mz > 1.5
        mz, intensity = mz[keep], intensity[keep]
    order = np.argsort(mz)
    return mz[order], intensity[order] / max(np.linalg.norm(intensity), 1e-12)

def cosine(a, b, tolerance=0.02):
    """Greedy one-to-one matching; all pairs within absolute m/z tolerance."""
    am, ai = a
    bm, bi = b
    if not len(am) or not len(bm):
        return 0.0
    left = np.searchsorted(bm, am - tolerance)
    right = np.searchsorted(bm, am + tolerance, side='right')
    pairs = [(ai[i] * bi[j], i, j) for i in np.flatnonzero(right > left)
             for j in range(left[i], right[i])]
    used_a, used_b, score = set(), set(), 0.0
    for value, i, j in sorted(pairs, reverse=True):
        if i not in used_a and j not in used_b:
            score += value
            used_a.add(i)
            used_b.add(j)
    return min(float(score), 1.0)

def validate_submission(frame, expected_ids):
    if list(frame.columns) != ['molecule_id', 'smiles'] or frame.empty:
        raise ValueError('Invalid submission columns or empty output')
    if frame.isna().any().any() or frame.molecule_id.duplicated().any():
        raise ValueError('Null values or duplicate molecule IDs')
    if set(frame.molecule_id) != set(expected_ids.molecule_id):
        raise ValueError('Submission molecule IDs differ from current test data')
    for value in frame.smiles:
        guesses = value.split(';')
        keys = [identity(s) for s in guesses]
        if not 1 <= len(guesses) <= 25 or any(x is None for x in keys):
            raise ValueError('Invalid SMILES or candidate count')
        if len({x[0] for x in keys}) != len(keys):
            raise ValueError('Duplicate scoring identities in a row')

_python_cosine = cosine
COMPILED_MATCHING = False
try:
    from numba import njit
    @njit(cache=False)
    def _fast_cosine(am, ai, bm, bi, tolerance):
        n, m = len(am), len(bm)
        values = np.empty(n*m, np.float64)
        aa = np.empty(n*m, np.int64)
        bb = np.empty(n*m, np.int64)
        count = 0
        for i in range(n):
            left = np.searchsorted(bm, am[i]-tolerance)
            right = np.searchsorted(bm, am[i]+tolerance, side='right')
            for j in range(left,right):
                values[count] = ai[i]*bi[j]; aa[count]=i; bb[count]=j; count+=1
        # Stable ascending order then reverse matches Python tuple ordering on ties.
        order = np.argsort(values[:count], kind='mergesort')
        used_a=np.zeros(n,np.bool_);used_b=np.zeros(m,np.bool_);score=0.0
        for k in range(count-1,-1,-1):
            v=order[k];i=aa[v];j=bb[v]
            if not used_a[i] and not used_b[j]:
                used_a[i]=True;used_b[j]=True;score+=values[v]
        return min(score,1.0)
    _fast_cosine(np.array([10.]),np.array([1.]),np.array([10.]),np.array([1.]),0.02)
    def cosine(a,b,tolerance=0.02):
        return float(_fast_cosine(a[0],a[1],b[0],b[1],tolerance))
    COMPILED_MATCHING = True
except Exception as error:
    print('Compiled matching unavailable; using tested Python matching:',type(error).__name__)
print('Compiled matching:',COMPILED_MATCHING,flush=True)
```

## 6. Build reference mass index

```python
# Use every supplied training source; no test labels or external structures.
all_smiles = set()
structure_counts = Counter()
for batch in reference.iter_batches(batch_size=100000, columns=['normalized_smiles']):
    df = batch.to_pandas()
    values = df['normalized_smiles'].dropna()
    values = values[values.map(lambda s: isinstance(s,str))]
    all_smiles.update(values)
    structure_counts.update(values)
print('Reference structures:', len(all_smiles), flush=True)
mass_records = []
for i,s in enumerate(sorted(all_smiles),1):
    if i%25000==0:print(f"Mass index: {i:,}/{len(all_smiles):,}",flush=True)
    mol = Chem.MolFromSmiles(s)
    if mol is not None and Chem.GetFormalCharge(mol) == 0 and '.' not in s:
        mass_records.append((Descriptors.ExactMolWt(mol), s))
mass_records.sort()
masses = np.array([r[0] for r in mass_records])
if not len(masses):
    raise ValueError('No neutral reference structures found')


print('Mass index complete',flush=True)

# A last-resort prior uses valid real training structures, never invented SMILES.
prior = []
for s, _ in structure_counts.most_common():
    ident = identity(s)
    if ident and ident[0] not in {x[0] for x in prior}: prior.append(ident)
    if len(prior) == 25: break
if not prior: raise ValueError('Reference library has no valid scoring identities')
```

## 7. Select candidate structures

```python
# Exact mass constrains the pool; a nearest-mass fallback is explicitly audited.
pools, details, smiles_to_queries = {}, [], {}
for mid in test_ids.molecule_id:
    target = float(neutral[mid])
    known_mass = np.isfinite(target) and target > 0
    tolerance = max(target * 15e-6, 0.003) if known_mass else 0.003
    lo, hi = np.searchsorted(masses, [target - tolerance, target + tolerance])
    ix = np.arange(lo, hi) if known_mass else np.array([], dtype=int)
    fallback = not len(ix)
    if fallback and known_mass:
        center = np.searchsorted(masses, target)
        nearby = np.arange(max(0, center - 200), min(len(masses), center + 200))
        ix = nearby[np.argsort(np.abs(masses[nearby] - target))[:100]]
    pool = {}
    for j in ix:
        mass, s = mass_records[j]
        key, canonical = s, s  # canonicalize only the ranked output candidates
        ppm = abs(mass - target) / target * 1e6
        pool.setdefault(key, {'smiles': canonical, 'ppm': ppm, 'spectra': {}})
        smiles_to_queries.setdefault(s, []).append((mid, key))
    if not pool:
        DIAGNOSTICS['prior_fallback_molecules'] += 1
        for key, canonical in prior:
            pool[canonical] = {'smiles': canonical, 'ppm': 0., 'spectra': {}}
    pools[mid] = pool
    details.append({'molecule_id': mid, 'neutral_mass': target,
                    'mass_fallback': bool(fallback), 'candidate_count': len(pool)})


print('Candidate pools:',len(pools),flush=True)
```

## 8. Prepare test spectra

```python
spectra = {}
for mid, rows in test.groupby('molecule_id', sort=False):
    spectra[mid] = [(str(r.spectrum_id), str(r.adduct), str(r.ionization_mode),
                     prepare_peaks(r.ms2_mzs, r.ms2_normalized_intensities, r.precursor_mz),
                     prepare_peaks(r.ms2_mzs, r.ms2_normalized_intensities, r.precursor_mz, True))
                    for r in rows.itertuples()]

print('Spectra ready',flush=True)
```

## 9. Compare reference spectra

```python
columns = ['normalized_smiles', 'adduct', 'ionization_mode',
           'precursor_mz', 'ms2_mzs', 'ms2_normalized_intensities']
compared = 0
for batch_no, batch in enumerate(reference.iter_batches(batch_size=32768, columns=columns)):
    df = batch.to_pandas()
    df = df[df.normalized_smiles.isin(smiles_to_queries)]
    for r in df.itertuples():
        if not np.isfinite(number(r.precursor_mz)):
            continue
        peaks = prepare_peaks(r.ms2_mzs, r.ms2_normalized_intensities, r.precursor_mz)
        losses = prepare_peaks(r.ms2_mzs, r.ms2_normalized_intensities, r.precursor_mz, True)
        for mid, key in smiles_to_queries[r.normalized_smiles]:
            evidence = pools[mid][key]['spectra']
            for sid, adduct, mode, query, qloss in spectra[mid]:
                if mode != mode_text(r.ionization_mode, r.adduct):
                    continue
                if adduct == adduct_text(r.adduct):
                    score = 0.85 * cosine(query, peaks) + 0.15 * cosine(qloss, losses)
                else:
                    score = 0.5 * cosine(qloss, losses)
                evidence[sid] = max(evidence.get(sid, 0.), score)
                compared += 1
    if batch_no % 10 == 0:
        print('Reference batch', batch_no, 'comparisons', compared, flush=True)
```

## 10. Rank and validate predictions

```python
rows, audit = [], []
for detail in details:
    mid = detail['molecule_id']
    ranked = []
    for key, c in pools[mid].items():
        support = np.array([c['spectra'].get(sid, 0.) for sid, *_ in spectra[mid]])
        spectral = 0.7 * support.max() + 0.3 * support.mean()
        score = spectral - 0.02 * min(c['ppm'] / 15., 100.)
        ranked.append((score, -c['ppm'], key, c['smiles'], spectral))
    ranked.sort(reverse=True)
    unique, seen = [], set()
    for item in ranked:
        ident = identity(item[3])
        if ident is None or ident[0] in seen: continue
        seen.add(ident[0]); unique.append((*item[:3],ident[1],item[4]))
        if len(unique) == 25: break
    if not unique:
        DIAGNOSTICS['final_identity_fallback'] += 1
        unique = [(0.,0.,key,smiles,0.) for key,smiles in prior]
    ranked = unique
    rows.append((mid, ';'.join(r[3] for r in ranked)))
    audit.append({**detail, 'top_score': ranked[0][0], 'top_spectral_score': ranked[0][4],
                  'top_mass_error_ppm': -ranked[0][1], 'n_guesses': len(ranked)})
submission = pd.DataFrame(rows, columns=['molecule_id', 'smiles'])
validate_submission(submission, test_ids)

print('Submission checks passed:',len(submission))
```
